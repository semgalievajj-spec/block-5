Голубой продукт: Демонстрирует классическую здоровую кривую удержания. Вначале наблюдается закономерный отток пользователей, но со временем график выходит на плато (примерно на уровне 40% к Day 4 – Day 7). Это означает, что у продукта есть ядро лояльной аудитории, которая регулярно продолжает им пользоваться.

Розовый продукт: Имеет катастрофическое падение показателей. Кривая стремительно падает вниз и полностью обнуляется к 5-му дню (Day 5), достигая 0%. Это свидетельствует о том, что продукт не представляет долгосрочной ценности для пользователей, в нем отсутствует вовлеченность или критически сломан пользовательский опыт (UX/onboarding).

Сравнение продуктов:
На старте (Day 0 и Day 1) показатели обоих продуктов примерно сопоставимы и находятся на высоком уровне.
Начиная со 2-го дня (Day 2), голубой продукт начинает значительно выигрывать по удержанию: на 2-й день у него удерживается около 53% пользователей, в то время как у розового — только около 40%.

```python
df['date'] = pd.to_datetime(df['date'])

november = df[
    (df['date'] >= '2023-11-01') &
    (df['date'] < '2023-12-01')
]

# Все уникальные пользователи за ноябрь
all_users = november['user_id'].nunique()

# Пользователи, посмотревшие хотя бы одно объявление
ad_users = november.loc[
    november['view_adverts'] > 0, 'user_id'
].nunique()

# Конверсия в процентах
conversion = ad_users / all_users * 100

print('Всего пользователей:', all_users)
print('Посмотрели объявления:', ad_users)
print(f'Конверсия: {conversion:.1f}%')
```

14. Scatter Plot
Это классический график для визуализации взаимосвязи двух непрерывных переменных $X$ и $Y$. По расположению точек прямо рассчитывается коэффициент корреляции (здесь видна сильная positive-корреляция).
    Line Plot
На данном графике изображены два временных ряда (две линии зеленых оттенков) во времени. По парным значениям этих двух линий в каждый момент времени можно рассчитать линейную или автокорреляцию.
    Correlation Heatmap
На этом графике коэффициенты корреляции между переменными $X$, $Y$ и $Z$ уже вычислены и прямо отображены числами и цветом.
